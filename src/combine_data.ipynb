```python
import pandas as pd
import os

import warnings
warnings.filterwarnings("ignore")
```

```python
DATA_PATH = "../../GISAID_database/"
os.listdir(DATA_PATH)
```

```python
dates =os.listdir(os.path.join(DATA_PATH, "dates_location"))
sorted(dates)
```

```python
dates_df = pd.concat(
    [pd.read_csv(os.path.join(DATA_PATH, "dates_location", fname), sep='\t') for fname in dates],
    ignore_index=True
)

dates_df.info()
```

```python
patients = os.listdir(os.path.join(DATA_PATH, "patient_metadata"))
sorted(patients)
```

```python
patients_df = pd.concat(
    [pd.read_csv(os.path.join(DATA_PATH, "patient_metadata", fname), sep='\t') for fname in patients],
    ignore_index=True
)
patients_df.info()
```

```python
df = pd.merge(dates_df[["Accession ID", "Submission date"]], patients_df, on="Accession ID", how="inner")
df.info()
```

```python
df.to_csv("../data/1Jan20_15Apr20_collected.csv", index=False)
```

```python
from tqdm import tqdm
entities = df2["Entity"].unique().tolist()
years = df2["Year"].unique().tolist()
data = []

for e in tqdm(entities, desc="extract yearly population"):
    entry = {"country": e}
    for y in years:
        entry[str(y)] = df2[(df2["Entity"] == e) & (df2["Year"] == y)]["all years"].values[0]
    data.append(entry)
df3 = pd.DataFrame(data)
df3.info()
```
